```python
from pydantic import BaseModel,Field
from typing import List,Annotated
class User(BaseModel):
    name:str
    age:int
```

```python
#Filed validation
class Product(BaseModel):
    title:str=Field(max_length=20)
    des:str=Field(min_length=20)
    price:int=Field(gt=0,le=200)

    
```

```python
product=Product(title="Samsung s23",des="This is amazing mobile from samsung",price=150)
product2=Product(title="iphone18",des="This is amazing mobile from apple",price=200)
```

```python
class Order(BaseModel):
    id:str=Field(min_length=6,max_length=12)
    product:Product
```

```python
order=Order(id="minjvn4544",product=product)
```

```python
order.product.des
```

```python
order.product.price
```

```python
class Carts(BaseModel):
    id:str=Field(min_length=6,max_length=12)
    items:List[Product]
    comment:str=Field(
        default="No comment"
    )
```

```python
cart=Carts(id="jsbjb444",items=[product,product2])
```

```python
cart
```

```python
for i in cart.items:
    print(i.title)
```

```python
#Annonated
commonValidator=Annotated[List[Product],"This is Common validator"]
```

```python
class Carts(BaseModel):
    id:str=Field(min_length=6,max_length=12)
    items:commonValidator
    comment:str=Field(
        default="No comment"
    )
```

```python
class Order(BaseModel):
    products:commonValidator
```

```python
ord=Order(products=[product,product2])
ord
```
