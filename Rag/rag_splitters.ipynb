```python
text="Pairing the weather with a sweet or playful text is a charming way to let someone know you're thinking of them. Whether it’s cozying up during a rainstorm, sending a caring reminder to stay warm on a chilly day, or sharing a bright note when the sun is out, a short weather-themed message adds a thoughtful touch to your day."
```

```python
#DocumentText Loader
from langchain_community.document_loaders import TextLoader,PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
loader=TextLoader("Stream(basic).txt")
texts=loader.load()
pdfloader=PyPDFLoader("AI_Vehicle_Management_System_Interview_Guide.pdf")
pdf=pdfloader.load()
splitter=RecursiveCharacterTextSplitter(chunk_size=500,chunk_overlap=100)
text_chunk=splitter.split_documents(texts)
len(text_chunk)
```
