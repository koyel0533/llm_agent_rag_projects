```python
from langchain_community.document_loaders import TextLoader
```

```python
#Text document loader
loader=TextLoader("Stream(basic).txt")
docs=loader.load() #loades the data in  from of list
docs 
oneDoc=docs[0]
oneDoc.page_content
```

```python
#pdf document loader
from langchain_community.document_loaders import PyPDFLoader
loader=PyPDFLoader("dbmsca2.pdf")
text=loader.load()
len(text) #5 pages
oneDoc=text[1]
print(oneDoc.page_content)
```

```python
#csv loader
from langchain_community.document_loaders import CSVLoader
loader=CSVLoader("IPL.csv")
docs=loader.load()
len(docs)
docs[:5]
```
