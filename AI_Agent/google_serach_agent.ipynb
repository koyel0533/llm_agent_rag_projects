```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_community.utilities import GoogleSerperAPIWrapper
import os
from langchain_groq import ChatGroq
from langchain.agents import create_agent
from langgraph.checkpoint.memory import MemorySaver
search=GoogleSerperAPIWrapper()
llm=ChatGroq(model="openai/gpt-oss-20b", groq_api_key=os.getenv("GROQ_API_KEY"))
#search.run("Which foreign city best for IT sector rather than India?")
agent=create_agent(model=llm, tools=[search.run], system_prompt="you are a agent can search for any question on google",checkpointer=MemorySaver())
question="what is the address of techsimplus lea"
response=agent.invoke({"messages":[{"role":"user","content":question}]}, {"configurable": {"thread_id": "1"}})

#len(response["messages"])
print(response["messages"][-1].content)

```
