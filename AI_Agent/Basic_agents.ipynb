```python
from langchain_core.tools import tool

@tool
def add_numbers(a: int, b: int) -> int:
    """Add two numbers together.
      Args:

    """
    return a + b
@tool
def multiply_numbers(a: int, b: int) -> int:
    """Multiply two numbers together.
      Args:

    """
    return a * b

add_numbers.invoke({"a": 5, "b": 3})
multiply_numbers.invoke({"a": 5, "b": 3})
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
import os
from langchain.agents import create_agent
llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash", google_api_key=os.getenv("GOOGLE_API_KEY"))
agent = create_agent(model=llm, tools=[add_numbers, multiply_numbers], system_prompt="You are a helpful assistant that can perform basic arithmetic operations.")
#llm_with_tool=llm.bind_tools([add_numbers])
#llm_with_tool.invoke('what is 123 plus 34.5?use the calculator tool')
#print('Text ',message.content)
#print('Tool Request',message.tool_calls)

response = agent.invoke({"messages":[{"role": "user", "content": "What is 3+5*8"}]})
print(response['messages'][-1].content)
'''for res in response['messages']:
    print(res)

    print('\n')
    '''
```
